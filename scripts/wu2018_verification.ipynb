```python
# Reconstruct Wu 2018's nuo-6 (ATFS-1-dependent) and atfs-1(et15,et17) gene sets from
# the real data, since the published counts (1,704 and 529) were never independently
# checked against the actual file. Method follows Soo & Van Raamsdonk 2021's own
# stated definition: "Genes upregulated in nuo-6 worms in an ATFS-1-dependent manner
# are genes that are upregulated in nuo-6 mutants but not nuo-6;atfs-1 mutants."
import os
if os.path.basename(os.getcwd()) == "scripts":
    os.chdir("..")

import pandas as pd

PATH = "data/raw/wu2018_AdditionalFile2.xlsx"

def load_sheet(name):
    return pd.read_excel(PATH, sheet_name=name, header=2)

nuo6 = load_sheet("nuo-6")
nuo6_atfs1 = load_sheet("nuo-6;atfs-1")
et15 = load_sheet("atfs-1(et15)")
et17 = load_sheet("atfs-1(et17)")

print(f"Sheet sizes (significant genes, both directions): nuo-6={len(nuo6)}, "
      f"nuo-6;atfs-1={len(nuo6_atfs1)}, et15={len(et15)}, et17={len(et17)}")

# nuo-6 ATFS-1-dependent upregulated
nuo6_up = set(nuo6[nuo6["logFC"] > 0]["ens_gene"])
nuo6_atfs1_up = set(nuo6_atfs1[nuo6_atfs1["logFC"] > 0]["ens_gene"])
nuo6_atfs1_dependent = nuo6_up - nuo6_atfs1_up

print(f"\nnuo-6 ATFS-1-dependent upregulated: {len(nuo6_atfs1_dependent)} "
      f"(published: 1,704 - {abs(len(nuo6_atfs1_dependent)-1704)} gene difference, "
      f"{100*len(nuo6_atfs1_dependent)/1704:.1f}% match)")
assert nuo6["ens_gene"].duplicated().sum() == 0 and nuo6_atfs1["ens_gene"].duplicated().sum() == 0, \
    "Duplicate gene IDs would explain a count mismatch - none found, so the gap above is real."

# atfs-1(et15) AND atfs-1(et17), both upregulated
et15_up = set(et15[et15["logFC"] > 0]["ens_gene"])
et17_up = set(et17[et17["logFC"] > 0]["ens_gene"])
both_up = et15_up & et17_up

print(f"atfs-1(et15) \u2229 atfs-1(et17) upregulated: {len(both_up)} (published: 529)")
assert len(both_up) == 529, "Expected an exact match to the published 529 figure."
print("Exact match confirmed.")
```
